## Tensor Basics

### Element-wise Addition and in-place addition, multiplication, division

```python
import torch
```

```python
x = torch.empty(2, 2, 3)
x = torch.rand(2, 2)
x = torch.zeros(2, 2, dtype=torch.float16)
x = torch.tensor([2.5, 0.1])
tensor_type =  x.dtype
size = x.size()
```

```python
x = torch.rand(2, 2)
y = torch.rand(2, 2)
print(x)
print(y)
z = x+y
z = torch.add(x, y)
y.add_(x)
```

### Slicing Operation

```python
x = torch.randint(0, 10, (5, 3))
print(x)
print(x[:, 0])
print(x[1, :])
print(x[1, 1].item())
# you can use .item() if there is only 1 value
```

### Reshaping tensor

```python
x = torch.randint(0, 10, (4, 4))
print(x)
y = x.view(16) # reshape into 1d vector
print(y)
z = x.view(-1, 8) # we can put -1, then specify other dimension for auto compute
print(z)
print(z.size())
```

### Converting from Numpy to Tensor

```python
import numpy as np
```

```python
a = torch.ones(5)
print(a)
print(type(a))
b = a.numpy()
print(b)
print(type(b))
```

```python
# If the two are in the same cpu, they will share the same memory location
a.add_(1)
print(a)
print(b)
```

```python
a = np.ones(5)
print(a)
b = torch.from_numpy(a)
print(b)

a += 1
print(a, b)
```

### GPU

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
if torch.cuda.is_available():
    x = torch.ones(5, device=device)
    y = torch.ones(5)
    y = y.to(device)
    z = x + y
    # z.numpy() causes error cause tensor in GPU cant be converted to numpy
    z.to("cpu")
    print(z)
```

### Requires_grad

```python
x = torch.ones(5, requires_grad=True)
print(x)
```

### Gradient Calculation with Autograd

```python
import torch
```

```python
x = torch.randn(3)
print(x)
```

```python
x = torch.randn(3, requires_grad=True)
# Now, if we do operations on this tensor, pytorch will save computational graph
print(x)

y = x+2
print(y)

z = y*y*2
z = z.mean()
print(z)

z.backward()
print(x.grad)
```
